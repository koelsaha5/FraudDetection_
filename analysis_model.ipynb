```python
# Feature Engineering and ML Model Training - 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,confusion_matrix
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer  # neu and cat sep
from sklearn.preprocessing import OneHotEncoder
```

```python
df.head()
```

```python
# Drop some columns 
df_model = df.drop(["nameOrig","nameDest","isFlaggedFraud"],axis=1)  # to drop and modify in df dataframe
```

```python
df_model.head()  # 9 cols
```

```python
categorical = ["type"]
numeric = ["amount","oldbalanceOrg","newbalanceOrig","oldbalanceDest","newbalanceDest"]
```

```python
# setting the target
y = df_model["isFraud"]                # y---> target var. = contains what the model wants to predict
X = df_model.drop("isFraud",axis = 1)  # X = feature var. = contains i/p fetures used by the model
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.3,stratify=y)
```

stratify=y ---> the dataset is highly imbalanced, used stratification so that the train and test sets preserve approximately *the same proportion of fraud and non-fraud transactions.*

```python
#preprocessing
preprocessor = ColumnTransformer(
    transformers = [
        ("num", StandardScaler(),numeric),
        ("Cat",OneHotEncoder(drop="first"),categorical)   # drop="first" as we want to reduce multicollinearity b/w independent features in ohe
    ],  # , necessary
    remainder = "drop"
)
```

ColumnTransformer because my dataset contains both numerical and categorical features, and they require different preprocessing. It lets me apply StandardScaler only to numerical columns and OneHotEncoder only to categorical columns.

```python
# creating ml pipeline - 1) preprocess (num and cat)  2) train logistic reg
pipeline = Pipeline([
    ("prep",preprocessor),
    ("clf",LogisticRegression(class_weight = "balanced",max_iter = 1000))
])
```

```python
# training the model
pipeline.fit(X_train,y_train)
```

```python
# making the prediction
y_pred = pipeline.predict(X_test)
```

```python
print(classification_report(y_test,y_pred))
```

we mainly evaluate the model using precision, recall and F1-score. Our recall is 94%, our model can detect 94% fraud trans out of all fraud trans, but precision is only 0.02 ---> the trans we predicted as fraud, only about 2% were actually fraud. 
f1 = 0.04 ---> shows the balance b/w precision and recall

```python
confusion_matrix(y_test,y_pred)
```
